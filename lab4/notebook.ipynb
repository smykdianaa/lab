```python
class MyAnimals:
    def __init__(self, animals, name, age, mass = None) -> None:
        print(f"В мене є тваринка {animals} яку звати {name} які {age} років")
        self.animals = animals
        self.name = name
        self.age = age
        self.mass = 0 if mass is None else mass
    
    def speak(self):
        if self.animals == "Кіт":
            return f"Моя Тваринка {self.animals} ({self.name}) говорить Мяууу"
        elif self.animals == "Собака":
            return f"Моя Тваринка {self.animals} ({self.name}) говорить Гавв"
        else:
            return f"Моя Тваринка {self.animals} ({self.name}) не розмовляє"
    
    def add_one_year(self):
        self.age += 1
        return f"Тваринка {self.name} ({self.animals}) прожила зі мною один рік. Новий вік: {self.age} років."
    
    def get_info(self):
        """Метод для виведення поточної інформації про об'єкт."""
        return f"-> {self.animals} на ім'я **{self.name}**, Вік: {self.age} років, Вага: {self.mass} кг."

a1 = MyAnimals("Рибки", "Неони", 1, 0.4)
a2 = MyAnimals("Кіт", "Мурка", 5, 10)
a3 = MyAnimals("Собака", "Рекс", 2, 15)

print(f"a1 (Неони): {a1.age} років")
print(f"a2 (Мурка): {a2.age} років")
print(f"a3 (Рекс): {a3.age} років")

print(a1.speak())
print(a2.speak())
print(a3.speak())

print(a1.add_one_year())
print(a2.add_one_year())
print(a3.add_one_year())

print(a1.get_info())
print(a2.get_info())
print(a3.get_info())
```

```python
class MyAnimals:
    total_animals = 0
    total_animal_types = set()

    def __init__(self, animals, name, age, mass = None) -> None:
        print(f"В мене є тваринка {animals} яку звати {name} які {age} років")
        self.animals = animals
        self.name = name
        self.age = age
        self.mass = 0 if mass is None else mass
    
        MyAnimals.total_animal_types.add(animals)
        MyAnimals.total_animals += 1

a1 = MyAnimals("Рибки", "Неони", 1)
MyAnimals("Рибки", "Гупії", 1)
MyAnimals("Рибки", "Соміки", 1)
MyAnimals("Кіт", "Мурка", 5, 10)

new_pets_data = [
    ("Собака", "Альфа", 3),
    ("Папуга", "Кеша", 10),
    ("Кіт", "Сніжок", 2) 
]
for animal_type, pet_name, pet_age in new_pets_data:
    MyAnimals(animal_type, pet_name, pet_age)


print(f"В мене зараз є тваринки {list(MyAnimals.total_animal_types)}, їх кількість {MyAnimals.total_animals}")


print(f"Я доступаюсь до змінної класу з обєкта {a1.total_animal_types}")

a1.total_animal_types = {"Собака"} 
print(f" Після присвоєння a1.total_animal_types = 'Собака'")
print(f"Я доступаюсь до змінної (тепер атрибута) з обєкта: {a1.total_animal_types}")
print(f"Я доступаюсь до змінної класу з класу: {MyAnimals.total_animal_types}")
```

```python
class MyAnimals:
    def __init__(self, animals, name, age) -> None:
        self.animals = animals
        self.name = name
        self.age = age
    
    def add_one_year(self):
        self.age += 1
        return f"Моя Тваринка {self.animals} ({self.name}) прожила за мною один рік. Тепер їй {self.age} років."

    @staticmethod
    def call_pet():
        """Статичний метод: не приймає 'self' і не залежить від стану об'єкта."""
        return "Тваринка біжить до нас"

a1 = MyAnimals("Рибки", "Гупії", 1)
a2 = MyAnimals("Кіт", "Мурка", 5)


print(f"Кличемо {a1.name}: {a1.call_pet()}")
print(f"Кличемо {a2.name}: {a2.call_pet()}")

print(f"Кличемо будь-яку тваринку через клас: {MyAnimals.call_pet()}")
```

```python
from datetime import datetime

class MyAnimals:
    max_age = 20
    
    def __init__(self, animals, name, age) -> None:
        self.animals = animals
        self.name = name
        self.age = age
    
    @property
    def live_untill(self):
        """Розраховує рік, до якого тварина теоретично доживе."""
        return datetime.today().year - self.age + self.max_age
    
    @property
    def info(self):
        """Нова властивість: повертає стислий опис тварини."""
        return f"Це {self.animals} на ім'я {self.name}, йому {self.age} років."
    
    def add_one_year(self):
        self.age += 1
        return f"Моя Тваринка {self.animals} прожила за мною один рік"

a1 = MyAnimals("Кіт", "Мурка", 5)
print(f"Інформація про тваринку: {a1.info}")
print(f"Мій {a1.animals} будже зі мною до {a1.live_untill} року")

a1.add_one_year()

print(f"Інформація про тваринку: {a1.info}")
print(f"Мій {a1.animals} будже зі мною до {a1.live_untill} року")
```

```python
class MyAnimals:
    total_csv = 0
    total_objects = 0

    def __init__(self, animals, name, age) -> None:
        self.animals = animals
        self.name = name 
        self.age = age
        MyAnimals.total_objects += 1

    @classmethod
    def from_csv(cls, csv_data: str):
        cls.total_csv += 1
        animal, name, age = csv_data.split(",")
        return cls(animal, name, int(age)) 
    
    def __init__(self, animals, name, age) -> None:
        self.animals = animals
        self._name = None  
        self.name = name   
        self.age = age
        MyAnimals.total_objects += 1

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, new_name):
        if not new_name.istitle():
            print(f"⚠️ Попередження: Ім'я '{new_name}' повинно починатися з великої літери. Виправлення: {new_name.capitalize()}")
            self._name = new_name.capitalize()
        else:
            self._name = new_name


a1 = MyAnimals.from_csv("Кіт,Мурка,5")

a2 = MyAnimals("Собака", "рекс", 2)

a3 = MyAnimals("Рибки", "гупії", 1) 

print(f"Мій {a1.animals}, звати {a1.name}, має {a1.age} років")
print(f"Мій {a2.animals}, звати {a2.name}, має {a2.age} років")
print(f"Мій {a3.animals}, звати {a3.name}, має {a3.age} років")

```
