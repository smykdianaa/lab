1. ### Інкапсуляція

```python
import random

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner  # публічний атрибут
        self.__balance = balance  # приватний атрибут

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Поповнено на: {amount} грн.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Знято: {amount} грн.")
            return amount
        else:
            print(f"Спроба зняти {amount} грн: Недостатньо коштів або невірна сума.")
            return "Insufficient funds"

    def get_balance(self):
        return self.__balance

account = BankAccount("Bohdan", 1000)
print(f"Початковий баланс {account.owner}: {account.get_balance()} грн.\n")

for i in range(1, 6):
    print(f"--- Транзакція №{i} ---")
    action = random.choice(['deposit', 'withdraw'])
    sum_value = random.randint(100, 1000) 
    
    if action == 'deposit':
        account.deposit(sum_value)
    else:
        account.withdraw(sum_value)

print("\n" + "="*25)
print(f"Кінцевий баланс: {account.get_balance()} грн.")
print("="*25)
```

2. ### Наслідування

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def display_info(self):
        return f"{self.brand} {self.model}"

    def start_engine(self):
        return f"Двигун автомобіля {self.brand} {self.model} успішно запущено! Врум-врум! 🏎️"

class Car(Vehicle):
    def __init__(self, brand, model, seats):
        super().__init__(brand, model)
        self.seats = seats

    def display_info(self):
        return f"{super().display_info()}, Кількість місць: {self.seats}"

car = Car("Toyota", "Camry", 5)

print(f"Інфо: {car.display_info()}")

print(car.start_engine())
```

3. ### Поліморфізм

```python
class Animal:
    def speak(self):
        return "..." 

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

class Fish(Animal):
    pass

animals = [Dog(), Cat(), Fish()]

for animal in animals:
    print(f"{animal.__class__.__name__}: {animal.speak()}")
```

### Використання парадигм для створення простої гри

```python
from abc import ABC, abstractmethod
from random import randint, choice

class Item(ABC):
    def __init__(self, name: str, health=500):
        self.name = name
        self.health = health
    
    @abstractmethod
    def attack(self, another_item):
        pass

    @abstractmethod
    def buff(self):
        """Метод для підсилення зброї"""
        pass

class Sword(Item):
    def __init__(self, name, attack_power: int):
        super().__init__(name=name)
        self.__attack_power = attack_power
        self._sharp = 0
    
    def attack(self, another_item: Item):
        current_attack = self.__attack_power + self._sharp + randint(0, 10)
        another_item.health -= current_attack
        return f"⚔️ Меч {self.name} завдає {current_attack} шкоди. У {another_item.name} залишилось: {max(0, another_item.health)} HP"
    
    def buff(self):
        self._sharp += 5
        return f"✨ Ви наточили меч! Гострота +5 (Зараз: {self._sharp})"

class Axe(Item):
    def __init__(self, name, attack_power: int):
        super().__init__(name=name)
        self.__attack_power = attack_power
    
    def attack(self, another_item: Item):
        current_attack = self.__attack_power + randint(0, 25)
        another_item.health -= current_attack
        return f"🪓 Сокира {self.name} вгатила на {current_attack} шкоди. У {another_item.name} залишилось: {max(0, another_item.health)} HP"

    def buff(self):
        bonus = randint(5, 15)
        self.__attack_power += bonus
        return f"💪 Лють! Базова атака сокири зросла на {bonus}!"

class Bow(Item):
    def __init__(self, name, attack_power: int, range_power: int):
        super().__init__(name=name)
        self.__attack_power = attack_power
        self.range_power = range_power

    def attack(self, another_item: Item):
        current_attack = self.__attack_power + randint(5, 15) + self.range_power
        another_item.health -= current_attack
        return f"🏹 Лук {self.name} вистрілив на {current_attack} шкоди. У {another_item.name} залишилось: {max(0, another_item.health)} HP"

    def buff(self):
        return self.reload()

    def reload(self):
        self.range_power += 2
        return f"⚙️ Перезарядка! Дальність зросла. (Зараз: {self.range_power})"

available_weapons = [
    Sword("Ескалібур", 40),
    Axe("Кривава Жатва", 35),
    Bow("Око Сокола", 30, 10)
]

player_weapon = choice(available_weapons)
enemy_weapon = choice([w for w in available_weapons if w != player_weapon])

print(f"🎮 ВАША ЗБРОЯ: {player_weapon.name} ({type(player_weapon).__name__})")
print(f"👾 ВАШ ВОРОГ: {enemy_weapon.name} ({type(enemy_weapon).__name__})\n")

round_num = 1
while player_weapon.health > 0 and enemy_weapon.health > 0:
    print(f"--- РАУНД {round_num} ---")
    print(f"Ваше HP: {player_weapon.health} | HP Ворога: {enemy_weapon.health}")
    
    choice_action = input("Виберіть дію: 1 - Атакувати, 2 - Підсилити зброю: ")
    
    if choice_action == "1":
        print(player_weapon.attack(enemy_weapon))
    else:
        print(player_weapon.buff())

    if enemy_weapon.health <= 0:
        print(f"\n🏆 ПЕРЕМОГА! {player_weapon.name} розгромив ворога!")
        break

    enemy_action = randint(1, 2)
    if enemy_action == 1:
        print(f"🤖 Ворог атакує: {enemy_weapon.attack(player_weapon)}")
    else:
        print(f"🤖 Ворог підсилюється: {enemy_weapon.buff()}")

    if player_weapon.health <= 0:
        print(f"\n💀 ПОРАЗКА! {enemy_weapon.name} виявився сильнішим.")
        break
    
    round_num += 1
    print("-" * 30)
```
