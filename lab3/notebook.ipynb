```python
import random

class Item:
    """Це наш клас який описує ітем в грі"""
    
    NAMES = ["Меч Сонця", "Сокира Воїна", "Спис Тіні", "Лук Ельфа", "Кинжал Змія"]
    
    def __init__(self):
        self._initial_durability = random.randint(50, 75)
  
        self.name = random.choice(self.NAMES)
        self.rarity = "epic"
        self.attack_power = random.randint(8, 12)
        self.durability = self._initial_durability
        self.magic_attribute = "Магія Вогню"
        self.description = f"Епічний {self.name}"
    
    @property
    def current_attack_value(self):
        """Property для читання значення атаки (аналог getter)."""
        return self.attack_power
    
    @property
    def durability_status(self):
        """Property для форматованого виведення міцності (аналог getter)."""
        return f"⚔️ {self.name} має міцність {self.durability}/{self._initial_durability}"


card = Item()

print(f"Початкова Атака: {card.attack_power} (прямий доступ) = {card.current_attack_value} (через property)")
print("-" * 60)

card.attack_power += 5

print(f"Атака після Бафу: {card.attack_power} (прямий доступ), {card.current_attack_value} (через property)")
print("-" * 60)

print(f"Статус Міцності: {card.durability_status}")

```

```python
import random

class Item:
    global_item_count = 0  
    """Це наш клас який описує ітем в грі"""
    def __init__(self):
        self._initial_durability = random.randint(30, 45)
        self.name = random.choice(["Меч Сонця", "Сокира Воїна", "Спис Тіні", "Лук Ельфа", "Кинжал Змія", "Поломаний Клинок"])

    def test_method(self):
        Item.global_item_count += 1
        return f"{self.name} має міцність {self._initial_durability} | Зміна лічильника: {Item.global_item_count}"
    
    @staticmethod
    def always_say_hello():
        return "Hello from Item class!"

i = Item()
b = Item()

print(type(i), type(b))

print(b.test_method(), i.test_method())

print(b.always_say_hello(), i.always_say_hello())
```

```python
import random

class CardGame:
    CARD_NAMES = ["Меч Сонця", "Сокира Воїна", "Спис Тіні", "Лук Ельфа", "Кинжал Змія"]
    
    def __init__(self):
        self.name = random.choice(self.CARD_NAMES)
        self.attack_power = random.randint(5, 15)
        self.health = random.randint(20, 30)
    
    def attack(self, other_card): 
        other_card.health -= self.attack_power
        return f" {self.name} атакує {other_card.name}. У {other_card.name} залишилось {other_card.health} здоров'я!"
    
    def defend(self):
        self.health += 2
        return f" {self.name} захищається, здоров'я відновилось до {self.health}!"
    
    def make_move(self, other_card):
        if random.choice([True, False]):
            return self.attack(other_card)
        else:
            return self.defend()


player_1 = CardGame()
print(f"**Наша карта:** {player_1.name} з атакою {player_1.attack_power} та здоров'ям {player_1.health}")
player_2 = CardGame()
print(f"**Ворожа карта:** {player_2.name} з атакою {player_2.attack_power} та здоров'ям {player_2.health}")

def shuffle_moves(player_1, player_2):
    if random.choice([True, False]):
        print(player_1.make_move(player_2))
        print(player_2.make_move(player_1))
    else:
        print(player_2.make_move(player_1))
        print(player_1.make_move(player_2))


print("\n--- ПОЧАТОК БИТВИ ---")
for i in range(1, 10):
    print(f"\nРаунд {i}:")
    shuffle_moves(player_1, player_2)
    
    if player_1.health <= 0:
        print(f"\n {player_1.name} програв! Переможець: {player_2.name}")
        break
    elif player_2.health <= 0:
        print(f"\n {player_2.name} програв! Переможець: {player_1.name}")
        break
```

```python
import random

class CardGame:
    CARD_NAMES = ["Меч Сонця", "Сокира Воїна", "Спис Тіні", "Лук Ельфа", "Кинжал Змія", "Поломаний Клинок"]
    
    def __init__(self):
        self.name = random.choice(self.CARD_NAMES)
        self.attack_power = random.randint(5, 15)
        self.health = random.randint(20, 30)
    
    def attack(self, other_card): 
        other_card.health -= self.attack_power
        return f" {self.name} (HP: {self.health}) атакує {other_card.name} (HP: {other_card.health + self.attack_power}). У {other_card.name} залишилось {other_card.health} здоров'я!"
    
    def defend(self):
        self.health += 2
        return f" {self.name} захищається, здоров'я відновилось до {self.health}!"
    
    def make_move(self, other_card):
        if random.choice([True, False]):
            return self.attack(other_card)
        else:
            return self.defend()

player_1 = [CardGame(), CardGame(), CardGame()]
print("--- Наші Карти (Гравець 1) ---")
for card in player_1:
    print(f"Карта: {card.name} | Атака: {card.attack_power} | Здоров'я: {card.health}")

player_2 = [CardGame(), CardGame(), CardGame()]
print("\n--- Ворожі Карти (Гравець 2) ---")
for card in player_2:
    print(f"Карта: {card.name} | Атака: {card.attack_power} | Здоров'я: {card.health}")

random.shuffle(player_1)
random.shuffle(player_2)

print("\n--- ПОЧАТОК БИТВИ ---")

for i in range(1, 15): 
    if len(player_1) == 0:
        print("\n Гравець 2 Переміг! Всі наші карти програли!")
        break
    elif len(player_2) == 0:
        print("\n Гравець 1 Переміг! Всі ворожі карти програли!")
        break
        
    print(f"\n===== РАУНД {i} =====")
    
    card1 = random.choice(player_1)
    card2 = random.choice(player_2)
    print(f"Грав. 1 ({card1.name}): {card1.make_move(card2)}")
    if card2.health > 0:
        print(f"Грав. 2 ({card2.name}): {card2.make_move(card1)}")

    if card1.health <= 0:
        print(f" Наша карта {card1.name} була знищена!")
        player_1.remove(card1)
        print(f"Залишилось карт у гравця 1: {len(player_1)}")
        
    if card2.health <= 0:
        print(f" Ворожа карта {card2.name} була знищена!")
        player_2.remove(card2)
        print(f"Залишилось карт у гравця 2: {len(player_2)}")
```

```python
class User:
    user_count = 0
    def __init__(self, name="Diana", surname="Smyk"):
        User.user_count += 1
        self.id = User.user_count
        self.name = name
        self.surname = surname
    
    def create_email(self):
        return f"{self.name.lower()}.{self.surname.lower()}@company.com"
    
    @property
    def full_info(self):
        email = self.create_email()
        return f"User #{self.id}: {self.name} {self.surname} ({email})"
    
    def save_to_file(self, filename="users.txt"):
        with open(filename, 'a', encoding='utf-8') as f:
            f.write(self.full_info + "\n")
        return f"Інформацію про {self.name} збережено у файл {filename}."

diana = User()
print(diana.save_to_file())
```
