Створити клас VowelCounter, який:

1. Приймає в конструкторі рядок
1. Має метод count(), який підраховує кількість голосних букв (a, e, i, o, u)
1. Має property vowel_count, яка повертає кількість голосних

Твоє завдання наступне:

1. Конструктор (__init__): приймає один рядок.

2. Метод count(): реалізуй логіку підрахунку латинських голосних букв (a, e, i, o, u). Підрахунок має бути нечутливим до регістру (враховувати і великі, і малі літери).

3. Property vowel_count: має повертати результат роботи методу count().

4. Метод __str__: має повертати рядок з текстом: "Виконав студент: [Прізвище Ім'я], група: [Номер групи]".

Демонстрація:

Створи список із 3-х різних текстових рядків.

У циклі for створи об'єкти класу для кожного рядка.

Для кожного об'єкта виведи в консоль результат методу __str__, сам текст та значення властивості vowel_count.

```python
class VowelCounter:
    def __init__(self, text):
        """Конструктор, що приймає рядок."""
        self.text = text
        self._vowels = "aeiou"

    def count(self):
        """Метод для обробки даних: підрахунок голосних (a, e, i, o, u)."""
        # Перетворюємо в нижній регістр для коректного підрахунку
        return sum(1 for char in self.text.lower() if char in self._vowels)

    @property
    def vowel_count(self):
        """Property для виведення результату підрахунку."""
        return self.count()

    def __str__(self):
        """Метод __str__ для виведення інформації про студента."""
        # Вкажіть тут своє Прізвище, Ім'я та Номер групи
        student_name = "Смик Діана"
        group_code = "КН-31"
        return f"Завдання виконав: {student_name}, група: {group_code}"

# --- Демонстрація роботи класу ---

# Список різних вхідних даних
test_strings = [
    "Hello World",
    "Python Programming is AMAZING",
    "Example text with many vowels"
]

# Створення об'єктів у циклі
for i, input_text in enumerate(test_strings, 1):
    counter = VowelCounter(input_text)
    
    # Виводимо інформацію про виконавця лише один раз на початку або для кожного об'єкта
    print(f"--- Об'єкт №{i} ---")
    print(counter) # Викликає метод __str__
    print(f"Вхідний текст: '{counter.text}'")
    print(f"Результат (vowel_count): {counter.vowel_count}")
    print("-" * 30)
```
